# Explore the training partition
Run training first. We inspect training data only, reserving validation for model selection and test for final evaluation.
Optional dependencies: `python -m pip install jupyterlab matplotlib`.

```python
from pathlib import Path
import sys
root = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
from src.config import load_config
from src.data.dataset import PointDataset
train = PointDataset(load_config(), "train")
print("Samples:", len(train))
print("Mean:", train.features.mean(0))
print("Class counts:", train.targets.bincount())
```

```python
import matplotlib.pyplot as plt
plt.scatter(train.features[:, 0], train.features[:, 1], c=train.targets, cmap="coolwarm", s=14)
plt.xlabel("Standardized x1")
plt.ylabel("Standardized x2")
plt.title("Training points: two nonlinear classes")
plt.axis("equal")
plt.show()
```
